```python
import numpy as np

X = np.array([[1,0]]) ##x1 and x2


W1 = np.array([
    [0.1,0.4,0.1],
    [0.5, 0.3, 0.7]
])

# print(w1.shape)

b1 = np.array([[0.1, 0.2, 0.3]])

# weighted sum
Z1 = np.dot(X, W1) + b1

np.dot(X, W1)


##activation function

def sigmoid(x):
    return 1/ (1+np.exp(-x))

A1 = sigmoid(Z1)


W2 = np.array([
    [0.8],
    [0.5],
    [0.9]
])

b2 = np.array([[0.2]])

Z2 = np.dot(A1, W2) + b2

A2 = sigmoid(Z2)

print("Hidden Layer Output:")
print(A1)

print("\nFinal Output:")
print(A2)






# NumPy performs all those multiplications and additions in one line using matrix multiplication.
```
