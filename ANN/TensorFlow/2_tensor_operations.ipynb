```python
import tensorflow as tf

A = tf.constant([
    [0,1],
    [1,2]
])

B = tf.constant([
    [5,6],
    [7,8]
])

print(A)
print(B)
```

```python
##operations:

#Addition
c = tf.add(A, B)
print(c)

#subtraction
d = tf.subtract(B,A)
print(d)

#multiplication
e = tf.multiply(A, B)
print(e)

f = tf.divide(B, A)
print(f)


# -------------------------------------------------------------------------------------
```

```python
#matrix multplication

p = tf.matmul(A, B)
print(p)



##reshape tensor
X = tf.constant([[1,2,3,4,5,6]])

y = tf.reshape(X, (2,3))
print(y)

##Transpose matrix (rows become column and vice versa)
z = tf.constant([
    [2,3],
    [4,5]
])
print(tf.transpose(z))
```

```python
#square the tensor

sq = tf.constant([[1,2,3,4,5,5]])
# print(tf.square(sq))

sqr = tf.constant([[1,2,3,4,5,5]], dtype=tf.float32)
# print(tf.sqrt(sqr))


#Reduce sum (calculate sum )

print(tf.reduce_sum(sq))

print(tf.reduce_mean(sq))
print(tf.reduce_min(sq))
print(tf.reduce_max(sq))
```
