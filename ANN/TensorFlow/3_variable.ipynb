```python
import tensorflow as tf

w = tf.Variable([1.0, 2.0, 3.0])

print(w)


w.assign([10.0, 20.0, 30.0])

print(w)
```

```python

# Update Only One Element
w = tf.Variable([1.0, 2.0, 3.0])

w[0].assign(100)

print(w)
```

```python
#Add to a variable
w = tf.Variable(5)

w.assign_add(10)

print(w)

# Subtract from a Variable
w.assign_sub(3)

print(w)
```

```python
# Mini Practical 
import tensorflow as tf

weight = tf.Variable(2.0)

learning_rate = 0.1

for epoch in range(5):
    gradient = 0.5

    weight.assign_sub(learning_rate * gradient)

    print(f"Epoch {epoch+1}: Weight = {weight.numpy()}")
```
