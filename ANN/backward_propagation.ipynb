```python
import numpy as np
import matplotlib.pyplot as plt


X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y = np.array([
    [0],[1],[1],[0]
])

W1 = np.array([
    [0.2, 0.4, 0.1],
    [0.5, 0.3, 0.7]
])

b1  =np.array([[0.1, 0.2, 0.3]])

W2 = np.array([
    [0.8],
    [0.5],
    [0.9]
])

b2 = np.array([[0.2]])

learning_rate = 0.1

def sigmoid(x):
    return 1/ (1+np.exp(-x))

##sigmoid derivatives

def sigmoid_derivative(x):
    return x*(1-x)

losses = []
epochs = 1000
for epoch in range(epochs):
    Z1 = np.dot(X,W1) + b1
    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)

    # print("prediction: ", A2)

    ##loss

    loss = 0.5 * np.square(y - A2)

    losses.append(np.mean(loss))


    ##backward porpagation

    dZ2 = (A2 -y) * sigmoid_derivative(A2)

    dW2 = np.dot(A1.T, dZ2)

    db2 = dZ2

    dZ1 = np.dot(dZ2, W2.T) * sigmoid_derivative(A1)

    dW1 = np.dot(X.T, dZ1)

    db1 = dZ1

##Update height

    W2 = W2 - learning_rate * dW2
    b2 = b2 - learning_rate * db2

    W1 = W1 - learning_rate * dW1
    b1 = b1 - learning_rate * db1



if epoch % 100 ==0:
    print(f"Epoch: {epoch}")
    print("Prediction: ", A2)
    print("Loss: ", loss)
    print("--------------------")


plt.plot(losses)
plt.title("Loss vs Epochs")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()

print("\n Updtated W2")
print(W2)

print("\n Updated W1")
print(W1)


```
